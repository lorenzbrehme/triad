```python
import os
from dotenv import load_dotenv
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_google_genai import GoogleGenerativeAIEmbeddings
load_dotenv(override=True)
os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY')
# embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-2")
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
# embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
# embeddings = HuggingFaceEmbeddings(model_name="BAAI/bge-small-en-v1.5")
# embeddings = HuggingFaceEmbeddings(
#     model_name="sentence-transformers/all-MiniLM-L6-v2"
# )
from langchain_postgres import PGVector
vector_store = PGVector(
    embeddings=embeddings,
    collection_name="my_docs",
    connection="postgresql+psycopg://postgres:password@localhost:5432/hotpot_embedding_01",
    # connection="postgresql+psycopg://postgres:password@localhost:5432/hotpot_sentence_transformer",
)
```

```python
import json
with open('../data/hotpot_dev_distractor_v1.json', 'r') as f:
    data_hotpot = json.load(f)
```

```python
# clear vector store
vector_store.delete_collection()
```

```python
# hotpotdata
counter = 0
collection = []
for item in data_hotpot:
    for context in item["context"]:
        # vector_store.add_texts([context[1]])
        context_str = ""
        for str in context[1]:
            context_str += str + " "
        counter += 1
        collection.append(context_str)
```

```python
# remove dublicates
collection = list(set(collection))
```

```python
print(counter)
print(len(collection))
```

```python
# hotpotdata
import time
counter = 0
batchsize = 64
batch = []
for item in collection:
        # print(item)
        counter += 1
        if len(batch) < batchsize:
            batch.append(item)
        else:
                batch.append(item)
                # try and wait for 60 seconds if error occurs
                for _ in range(5):
                    try:
                        vector_store.add_texts(batch)
                        break
                    except Exception as e:
                        print(f"Error occurred: {e}. Retrying in 60 seconds...")
                        time.sleep(60)
                batch = []
        
        print(f"Inserted {counter} documents", end="\r")
vector_store.add_texts(batch)
print(f"\nTotal inserted documents: {counter}")
        
```
