```python
import json
data =[]
with open('../data/musique_ans_v1.0_dev.jsonl', 'r') as file:
    for line in file:
        data_line = json.loads(line)
        data.append(data_line)
```

```python
# get data structure
print(f"Loaded {len(data)} records")
# gat all paramters
all_params = set()
for item in data:
    for key in item.keys():
        all_params.add(key)
        
print(f"All parameters: {all_params}")
# get parameters of paragraph
paragraph_params = []
for item in data:
    if 'paragraphs' in item:
        len_paragraphs = len(item['paragraphs'])
        paragraph_params.append(len_paragraphs)
        
#avarage number of paragraphs
avg_paragraphs = sum(paragraph_params) / len(paragraph_params) if paragraph_params else 0
print(f"Average number of paragraphs: {avg_paragraphs}")
#list how often a paragraph has n elements
new_data = []
for item in data:
    if 'paragraphs' in item:
        length = len(item['paragraphs'])
        if length == 20:
            new_data.append(item)
data = new_data
new_data = []
for item in data:
    if 'paragraphs' in item:
        counter = 0
        for paragraph in item['paragraphs']:
           if paragraph['is_supporting'] == True:
               counter += 1
        if counter == 2:
            new_data.append(item)
data = new_data
data_final = []
for data_item in data[0:600]:
    contextlist = []
    supporting_context_list = []
    unsupporting_context_list = []
    for paragraph in data_item['paragraphs']:
        contextlist.append(paragraph['paragraph_text'])
        if paragraph['is_supporting'] == True:
            supporting_context_list.append(paragraph['paragraph_text'])
        if paragraph['is_supporting'] == False:
            unsupporting_context_list.append(paragraph['paragraph_text'])
    
        
    data_final.append({"question_id": data_item['id'],
                        "multi_hop_question": data_item['question'],
                        "multi_hop_answer": data_item['answer'],
                        "contextList": contextlist,
                        "supporting__sentences": supporting_context_list,
                        "not_supporting_sentences": unsupporting_context_list,
                        
                       })
print(f"Final data length: {len(data_final)}")

with open('./static_qas/musique_preprocessed_neu.json', 'w') as f:
    json.dump(data_final, f, indent=4)
```
